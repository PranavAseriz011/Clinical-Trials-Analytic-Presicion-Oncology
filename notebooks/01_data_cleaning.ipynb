# 🔧 Data Type Inspection

Before performing transformations, the data types of all columns are inspected to identify columns that require conversion into appropriate formats.

```python
df["enrollment_count"].describe()
```

```python
df["start_date"] = pd.to_datetime(df["start_date_raw"], errors="coerce")

df["completion_date"] = pd.to_datetime(df["completion_date_raw"], errors="coerce")
```

```python
df[["start_date_raw", "start_date",
    "completion_date_raw", "completion_date"]].head()
```

## 📌 Data Type Observations

- Most columns were imported as text (`object`) from the CSV file.
- The `enrollment_count` column was correctly recognized as a numeric (`float64`) column.
- The date columns contained mixed formats (`YYYY-MM` and `YYYY-MM-DD`), so they were converted using `pd.to_datetime()` with `format="mixed"` to correctly parse both formats.

```python
df["start_date"] = pd.to_datetime(
    df["start_date_raw"],
    format="mixed",
    errors="coerce"
)

df["completion_date"] = pd.to_datetime(
    df["completion_date_raw"],
    format="mixed",
    errors="coerce"
)
```

```python
df[[
    "start_date_raw",
    "start_date",
    "completion_date_raw",
    "completion_date"
]].head()
```

# 🧹 Text Standardization

This step removes unnecessary whitespace from text columns to improve consistency and prevent duplicate values caused by formatting differences.

```python
df.select_dtypes(include = "object").columns
```

```python
text_columns = df.select_dtypes(include = "object").columns
```

```python
df[text_columns] = df[text_columns].apply(
    lambda x: x.str.strip() if x.dtype == "object" else x
)
```

```python
df.head()
```

## 📌 Observation

Leading and trailing whitespaces were removed from all text columns to ensure consistent formatting and avoid mismatched values during analysis.

# 🔍 Missing Value Investigation

Missing values were analyzed column by column to understand whether they represent missing information, non-applicable values, or expected characteristics of clinical trial data.


```python
df["overall_status"].value_counts() 
```

## 📌 Observation

The `overall_status` column contains multiple trial states such as **Completed**, **Recruiting**, **Terminated**, and **Suspended**.

The high number of missing values in the `why_stopped` column is expected because trials that are completed or still ongoing do not require a stopping reason. Therefore, missing values in `why_stopped` will be retained rather than treated as data quality issues.

```python
df["phase_raw"].value_counts(dropna=False)
```

```python
df[df["phase_raw"].isna()]["study_type"].value_counts()
```

```python
df[df["phase_raw"] == '["NA"]']["study_type"].value_counts()
```

## 📌 Observation

Investigation of the `phase_raw` column showed two different representations of unavailable phase information:

- `NaN` values were primarily associated with **Observational** studies, where clinical phases are generally not applicable.
- The string value `["NA"]` was found in **Interventional** studies and was retained because it represents an explicit value present in the source dataset rather than a missing entry.

Therefore, no modifications were made to the `phase_raw` column during data cleaning.

```python
df["sex"].value_counts(dropna=False)
```

## 📌 Observation

The `sex` column contains only 17 missing values out of 42,829 records (approximately 0.04% of the dataset).

Since the correct participant eligibility cannot be inferred, these missing values were retained to preserve data integrity.

```python
df["minimum_age"].value_counts(dropna=False)
```

```python
df["maximum_age"].value_counts(dropna=False)
```

## 📌 Observation

The `minimum_age` column contains a small number of missing values, while the `maximum_age` column has a large proportion of missing entries.

These missing values were retained because they likely represent studies where age limits were not specified rather than errors in the dataset. No imputation was performed to avoid introducing incorrect information.

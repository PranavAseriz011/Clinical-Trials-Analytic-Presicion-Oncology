```python
import pandas as pd
import sys

sys.path.append("../")

from bioinformatics.biomarker_detector import (
    detect_biomarkers,
    detect_alterations,
    match_gene_alterations
)
```

```python
df = pd.read_csv("../Data/trials_master_CLEANED.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
text = df.loc[0, "eligibility_criteria"]

print(text)
```

```python
print("Biomarkers:")
print(detect_biomarkers(text))

print("\nAlterations:")
print(detect_alterations(text))

print("\nGene-Alteration Matches:")
print(match_gene_alterations(text))
```

```python
df["detected_biomarkers"] = df["eligibility_criteria"].apply(detect_biomarkers)

df["detected_alterations"] = df["eligibility_criteria"].apply(detect_alterations)

df["gene_alteration_matches"] = df["eligibility_criteria"].apply(match_gene_alterations)
```

```python
print("Biomarker-positive trials:",
      df["detected_biomarkers"].apply(len).gt(0).sum())

print("Alteration-positive trials:",
      df["detected_alterations"].apply(len).gt(0).sum())

print("Gene-alteration matches:",
      df["gene_alteration_matches"].apply(len).gt(0).sum())
```

```python
output_path = "../Data/trials_master_ANNOTATED.csv"

df.to_csv(output_path, index=False)

print("Saved successfully!")
print(output_path)
```
